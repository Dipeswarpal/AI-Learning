```python
# Encapsulation
# Encapsulation means wrapping data and methods together and controlling access

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

account = BankAccount(1000)
account.deposit(500)
print(account.balance)
```

```python
# Name mangling in Python is a technique used to make class attributes and methods private by adding a prefix of double underscores (__) to their names. This helps prevent accidental access or modification of these attributes from outside the class.

# Private variables
# ------------------
# python uses _ conventions.
# protected
    # self._salary
# private
    # self.__salary

class Employee:
    def __init__(self):
        self.__salary = 50000  # Private variable

emp = Employee()
# print(emp.__salary)  # This will raise an AttributeError

# Name mangling
print(emp._Employee__salary)  # Accessing the private variable using name mangling
```

```python
# Inheritance
# Inheritance allows a class to inherit attributes and methods from another class.

class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

dog = Dog()
dog.eat()  # Inherited method from Animal class
dog.bark()  # This will raise an AttributeError since bark() is not defined in Dog class
```

```python
# Multiple Inheritance

class Father:
    def skill(self):
        print("Driving")

class Mother:
    def hobby(self):
        print("Music")

class Child(Father, Mother):
    pass

c = Child()
c.skill()
c.hobby()
```

```python
# Method Overriding
# Child class changes parent behavior by defining a method with the same name as in the parent class.

class Animal:
    def sound(self):
        print("Some sound")

class Dog(Animal):
    def sound(self):
        print("Bark")
        super().sound()  # Call the parent class method

dog = Dog()
dog.sound()  # This will call the overridden method in Dog class
```

```python
class Person:
    college = "ABC Institute"
    total_records = 0

    def __init__(self, name, phone, city):
        self.name = name
        self.phone = phone
        self.city = city
        Person.total_records += 1

    def display(self):
        print("\n____________________________")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")
        print(f"City: {self.city}")
        print(f"College: {Person.college}")
```

```python
class Student(Person):
    def __init__(self, name, phone, city, roll, mark):
        super().__init__(name, phone, city)
        self.roll = roll
        self.mark = mark

    def display(self):
        super().display()
        print(f"Roll: {self.roll}")
        print(f"Mark: {self.mark}")
```

```python
class NonTechEmployee(Person):
    def __init__(self, name, phone, city, empid, desig, salary):
        super().__init__(name, phone, city)
        self.empid = empid
        self.desig = desig
        self.salary = salary

    def display(self):
        super().display()
        print(f"Employee ID: {self.empid}")
        print(f"Designation: {self.desig}")
        print(f"Salary: {self.salary}")
```

```python
class TechEmployee(NonTechEmployee):
    def __init__(self, name, phone, city, empid, desig, salary, sub1):
        super().__init__(name, phone, city, empid, desig, salary)
        self.sub1 = sub1

    def display(self):
        super().display()
        print(f" Technical Subject: {self.sub1}")
```

Abstract Method

```python
from abc import ABC, abstractmethod

# Abstract Class
class Animal(ABC):
    # Abstract method
    @abstractmethod
    def sound(self):
        pass

    # Non-abstract method
    def sleep(self):
        print("Animal is sleeping")

class Dog(Animal):

    # implementing the abstract method
    def sound(self):
        print("Dog barks")


d = Dog()
d.sound()
d.sleep()
```

```python
import sqlite3

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

cursor. execute("""
        CREATE TABLE IF NOT EXISTS employee(
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT,
            salary REAL
        )
""")

connection.commit()
connection.close()
```

```python
# Inserting Records

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

cursor . execute(
    "INSERT INTO employee (name, salary) VALUES ('Anand',50000)"
)

connection.commit()
connection.close()
```

Using Placeholders (Recommended)
Instead of

cursor.execute(
"INSERT INTO employee VALUES (101, 'Anand',50000)"
)

Use
cursor.execute(
"INSERT INTO employee VALUES (?,?,?)",
(102,"Rahul",60000)|
)

Why?
-Prevents SQL Injection
-Easier to reuse
-Corporate standard

```python
# Inserting Multiple Records

employees = [
    ("Priya",55000),
    ("Kumar",62000),
    ("Divya",58000)
]

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

cursor .executemany (
"INSERT INTO employee (name, salary) VALUES (?,?)",
employees
)

connection.commit()
connection.close()
```
