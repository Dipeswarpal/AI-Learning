```python
num = int(input("Enter any Number from 1-3:"))

match num:
    case 1 :
        print("One")
    case 2 :
        print("Two")
    case 3 :
        print("Three")
    case _:
        print("Invalid Number")
```

```python
ch = str(input("Enter a character: "))[0:1].lower()

match ch:
    case 'a' | 'e' | 'i' | 'o' | 'u':
        print(f"{ch} is a vowel")
    case _:
        print(f"{ch} is not a vowel")
```

```python
# default Parameters
def greet(name="Anand"):
    print("Welcome", name)

greet()
greet("Dipeswar")
```

```python
# Keyword Arguments
def employees(name,age):
    print(name, age)
employees(name="Dipeswar", age=23)
```

```python
# Arbitory Arguments
# Used when the number of arguments is unknown.

def total(*numbers):
    print(numbers)

total(10, 20, 30, 40)
total(10, 20, 30, 40, 50, 60, 70)
```

```python
# Keyword Arbitrary Arguments(**kwargs)

def employees(**details):
    print(details)

employees(name="Dipeswar", age=23, department="IT", location="Bangalore")
```

```python
# Return Multiple Values

def calculate(a, b):
    return a + b, a - b

res1,res2 = calculate(10, 5)

print(f"Sum = {res1}, Diff = {res2}")
```

```python
def finance(basicpay):
    DA = 0.1 * basicpay
    HRA = 0.2 * basicpay
    totalsalary = DA + HRA + basicpay
    return  DA, HRA, totalsalary

bp = float(input("Enter basic pay: "))
print(finance(bp))
```

```python
# Class with attributes (Instance Variables)

class Student:
    def __init__(self, roll, name, mark):
        self.roll = roll
        self.name = name
        self.mark = mark

    def setCourse(self, course):
        self.course = course
```

```python
# s1 = Student()
s1.setCourse(101, 'Anand', 56.7)
s1.setCourse("BTech CSE")
print(s1.roll)
print(s1.name)
print(s1.mark)
print(s1.course)
```
